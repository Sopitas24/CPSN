```python

08 Visualización para ciencia de datos - Arreglar el gráfico (Core).txt
100%

import pandas as pd
import matplotlib.pyplot as plt
from io import StringIO

data = """fecha,ventas
2023-01-01,200
2023-01-02,220
2023-01-03,210
2023-01-04,205
2023-01-05,215
2023-01-06,225
"""



df = pd.read_csv(StringIO(data))

df["fecha"] = pd.to_datetime(df["fecha"], format="%Y-%m-%d")


df = df.sort_values("fecha")

print(df)

plt.figure(figsize=(10, 6))
plt.plot(df["fecha"], df["ventas"], marker="o", color="steelblue")
plt.title("Ventas diarias")
plt.xlabel("Fecha")
plt.ylabel("Ventas")
plt.xticks(df["fecha"], rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 6))
plt.scatter(df["fecha"], df["ventas"], color="darkorange", s=70)
plt.title("Ventas diarias: gráfico de dispersión")
plt.xlabel("Fecha")
plt.ylabel("Ventas")
plt.xticks(df["fecha"], rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 6))
plt.hist(df["ventas"], bins=5, color="skyblue", edgecolor="black")
plt.title("Distribución de las ventas diarias")
plt.xlabel("Ventas")
plt.ylabel("Frecuencia (cantidad de días)")
plt.yticks(range(0, len(df) + 1))
plt.tight_layout()
plt.show()
Mostrando 08 Visualización para ciencia de datos - Arreglar el gráfico (Core).txt
```
